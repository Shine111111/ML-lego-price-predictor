```python
import pandas as pd

#loading the data
df = pd.read_csv('lego_data.csv')

print(df.head())
```

```python
#I have to convert text categories into numbers 1s and 0s
#This created new columns like Theme_Star_Wars, Theme_Ninjago, etc.
df_encoded = pd.get_dummies(df, columns=['Theme'])

#Our Inputs are in the X axis and they are all the columns except MSRP, which is our target variable(y)
X = df_encoded.drop('MSRP', axis=1)
#y is only the msrp
y = df_encoded['MSRP']

print("Features ready for training!")
```

```python
from sklearn.linear_model import LinearRegression

#Initialzing the model
model = LinearRegression()

#Training the model on my lego data
model.fit(X, y)

print("Model trained successfully!")
```

```python
#Now I can use the model to make predictions on new lego sets
new_set = pd.DataFrame({
    'Pieces':[1200],
    'Minifigures': [5],
    'Theme_City':[0],
    'Theme_Creator':[0],
    'Theme_Ideas':[0],
    'Theme_Ninjago':[0],
    'Theme_Star Wars':[1],
    'Theme_Technic':[0]
})

predicted_price = model.predict(new_set)

print(f"The predicted price for the new lego set is: ${predicted_price[0]:.2f}")
```
